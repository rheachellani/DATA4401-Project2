```python
# evaluation metrics and hyperparameters
lossfunc = nn.BCEWithLogitsLoss()
lr = 1e-6
epochs = 1000
optim = torch.optim.SGD(model.parameters, lr=lr)
```

```python
def train(model, hyperparams, train):
    for t in range(hyperparams.epochs):
        y_pred = model(train.x)

        loss = hyperparams.lossfn(y_pred, train.y)
        if t % 100 == 99:
            print(t, loss.item())

        hyperparams.optimizer.zero_grad()

        loss.backward()

        hyperparams.optimizer.step()
```

```python
def evaluate(model, test):
    preds = model(test.x)
    targets = test.y
    
    # 2. Compute ROC curve metrics (FPR, TPR, thresholds)
    b_roc = BinaryROC()
    fpr, tpr, thresholds = b_roc(preds, targets)

    # 3. Compute AUROC score
    b_auroc = BinaryAUROC()
    auroc_score = b_auroc(preds, targets)

    # 4. Plot the ROC Curve
    # Convert tensors to numpy arrays for plotting
    plt.figure(figsize=(6, 5))
    plt.plot(fpr.numpy(), tpr.numpy(), color='darkorange', lw=2, 
             label=f'ROC curve (AUC = {auroc_score.item():.4f})')
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--') # Diagonal random line
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate (FPR)')
    plt.ylabel('True Positive Rate (TPR)')
    plt.title('Receiver Operating Characteristic (ROC) Curve')
    plt.legend(loc="lower right")
    plt.grid(True)
    plt.show()
```

```python
for model in models:
    for hyperparams in hyperparamset:
        train(model, hyperparams, train)
        evaluate(model, test)
```
